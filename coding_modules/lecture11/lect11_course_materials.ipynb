```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("CoreDataNY10.csv")
```

```python
plt.figure(figsize=[5,4],dpi=200)
plt.scatter(df['Ti'],df['DepthFromTrenton_ft'],s=5,c='gray')
plt.xlabel("Titanium Abundance (ppm)")
plt.ylabel("Depth from Trenton (ft)")
plt.show()
```

```python
df_rolling_mean = df[['DepthFromTrenton_ft','Ti']].rolling(window=10).mean().dropna()
```

```python
plt.figure(figsize=[5,4],dpi=200)
plt.scatter(df['Ti'],df['DepthFromTrenton_ft'],s=5,c='gray')
plt.plot(df_rolling_mean['Ti'],df_rolling_mean['DepthFromTrenton_ft'],c='red')
plt.xlabel("Titanium Abundance (ppm)")
plt.ylabel("Depth from Trenton (ft)")
plt.show()
```

# period

```python
x = np.arange(0,1000)
y = x % 185
z = (x + 60) % 185
```

```python
plt.figure(figsize=[8,4],dpi=200)
plt.plot(x,y,c='dodgerblue',ls='--', label='y = x % 185')
plt.plot(x,z,c='coral',ls='-', label='z = (x + 60) % 185')
plt.ylim(-30,240)
plt.legend()
```
